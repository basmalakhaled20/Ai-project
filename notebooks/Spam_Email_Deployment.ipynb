```python
!pip install streamlit
import streamlit as st
import pickle
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
import nltk
import string

nltk.download('stopwords')
```

```python
model = pickle.load(open(r"/content/naive_bayes_model.pkl", "rb"))
vectorizer = pickle.load(open(r"/content/vectorizer.pkl", "rb"))
label_encoder = pickle.load(open(r"/content/label_encoder.pkl", "rb"))
```

```python
st.title(" Spam Mail Detection")
```

```python
mail = st.text_input("Enter your message:")
```

```python
if st.button("confirm"):
    if mail.strip() == "":
        st.warning("Please type a message first!")
    else:

        stemmer = PorterStemmer()
        stopwords_set = set(stopwords.words('english'))
        text = mail.lower()
        text = text.translate(str.maketrans('', '', string.punctuation)).split()
        text = [stemmer.stem(w) for w in text if w not in stopwords_set]
        cleaned = " ".join(text)


        X = vectorizer.transform([cleaned])
        result = model.predict(X)[0]

        label_name = label_encoder.inverse_transform([result])[0]

        if str(label_name).lower() == "spam":
            st.error("✖ This is a Spam message.")
        else:
            st.success("✔ This is a Ham message.")
```
